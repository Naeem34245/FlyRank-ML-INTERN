## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research question:** Which content pages should FlyRank's review team look at first 
when they have limited time to act on content refresh?

**Unit of analysis:** one page (one content_id) at one point in time.

**Decision it improves:** which pages a content reviewer opens first, out of hundreds 
of candidates, given limited review capacity.

**Action someone takes:** a reviewer opens the top-ranked pages and decides to refresh, 
expand, protect, prune, or monitor each one, guided by the reason codes attached to it.

**Cost of a wrong recommendation:** 
- False positive (page flagged but not actually a problem): reviewer wastes limited time 
  on a page that didn't need attention, while a real problem page waits.
- False negative (a real declining/stale page not surfaced): the page keeps losing 
  visibility or traffic silently, and no one notices until it's much worse.

**Why data/ML helps here (not just intuition):** with tens of thousands of pages and only 
a handful of reviewer-hours per week, no human can manually check every page. A ranked, 
evidence-backed queue turns an impossible manual task into a short, prioritized list — 
and a learned model can combine many weak signals (position, CTR, freshness, engagement) 
in ways a single fixed rule can't.
